```python
import os
import torch
from ultralytics import YOLO

# Eğitilen modelin ağırlık yolu
model_path = os.path.abspath("../models/yolo_runs/flir_yolov8n/weights/best.pt")
model = YOLO(model_path)

# Doğrulama kümesinde değerlendirme
metrics = model.val(data="../dataset/yolo/data.yaml", split='val', workers=0)

print("\n" + "="*40)
print("       YOLOv8 NESNE TESPİTİ BAŞARIMI")
print("="*40)
print(f"mAP@50      : %{metrics.box.map50 * 100:.2f}")
print(f"mAP@50-95   : %{metrics.box.map * 100:.2f}")
print(f"Precision   : %{metrics.box.mp * 100:.2f}")
print(f"Recall      : %{metrics.box.mr * 100:.2f}")
print("="*40)
```

```python
import matplotlib.pyplot as plt
from PIL import Image

run_dir = "../models/yolo_runs/flir_yolov8n"

# Eğitim süreci grafikleri ve karışıklık matrisi
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

results_img = Image.open(f"{run_dir}/results.png")
axes[0].imshow(results_img)
axes[0].set_title("Eğitim ve Doğrulama Kayıp / Metrik Eğrileri")
axes[0].axis('off')

cm_img = Image.open(f"{run_dir}/confusion_matrix.png")
axes[1].imshow(cm_img)
axes[1].set_title("YOLOv8 Karışıklık Matrisi")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import glob

# Örnek test görselleri
val_images = glob.glob("../dataset/yolo/images/val/*.jpg")[:4]

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
axes = axes.flatten()

for idx, img_path in enumerate(val_images):
    # Model çıkarımı (conf eşiği 0.30)
    results = model.predict(source=img_path, conf=0.30, verbose=False)
    plotted_img = results[0].plot()  # Kutuları çizilmiş RGB görsel
    
    axes[idx].imshow(plotted_img)
    axes[idx].set_title(f"Örnek Termal Kare {idx+1}")
    axes[idx].axis('off')

plt.tight_layout()
plt.show()
```
